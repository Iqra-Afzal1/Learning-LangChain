```python
from langchain_huggingface import ChatHuggingFace, HuggingFaceEndpoint
from pydantic import BaseModel
from typing import Annotated
from dotenv import load_dotenv
import os
```

```python
load_dotenv(override=True)
```

```python
class Response(BaseModel):

    summary : Annotated[str, "Must be between 2 to 5 lines"]
    sentiment : Annotated[str, "Must be single word"]
```

```python
class ConfigureModel:

    def __init__(self, model_name:str = "Qwen/Qwen2.5-7B-Instruct"):
        self.model_name = model_name
        self.model = None
        self._initiate_model()

    def _initiate_model(self):
        print(f"Trying to connected with model {self.model_name}...")
        try:
            llm = HuggingFaceEndpoint(
                repo_id = self.model_name,
                task = 'text-generation',
                huggingfacehub_api_token=os.getenv('HUGGINGFACEHUB_ACCESS_TOKEN')
            )
            self.model = ChatHuggingFace(llm = llm)
            self.model = self.model.with_structured_output()
            print("Connection with model successful !")

        except Exception as e:
            print(f"Error while connecting with model...\n {e}")

    def get_response(self, prompt):
        response = self.model.invoke(prompt)

        return response
```
