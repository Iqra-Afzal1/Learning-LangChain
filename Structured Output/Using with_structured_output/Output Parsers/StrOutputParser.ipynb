```python
from langchain_huggingface import ChatHuggingFace, HuggingFaceEndpoint
import os
from dotenv import load_dotenv
from langchain_core.output_parsers import StrOutputParser
from langchain_core.prompts import PromptTemplate
```

```python
load_dotenv(override=True)
```

- StrOutputParser extracts model response and returns us string
- It's handy when using chains

```python
class ConfigureModel:

    def __init__(self, model_name:str = "Qwen/Qwen2.5-7B-Instruct"):
        self.model_name = model_name
        self.model = None
        self._initiate_model()

    def _initiate_model(self):
        try:
            print(f"Trying to connect with model {self.model_name}...")
            llm = HuggingFaceEndpoint(
                repo_id = self.model_name,
                task = 'text-generation',
                huggingfacehub_api_token=os.getenv('HUGGINGFACEHUB_ACCESS_TOKEN')
            )
            self.model = ChatHuggingFace(llm = llm)
            print("Connection successful with model !")

        except Exception as e:
            print(e)

    def get_response(self, prompt):
        print("Query sent to model, waiting for response...")
        try:
            response = self.model.invoke(prompt)
            return response.content

        except Exception as e:
            print(e)
```

```python
model_configurer = ConfigureModel()
```
