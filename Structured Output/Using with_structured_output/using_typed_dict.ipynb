```python
from langchain_huggingface import ChatHuggingFace, HuggingFaceEndpoint
from typing import TypedDict
from dotenv import load_dotenv
import os
```

```python
load_dotenv(override=True)
```

```python
class Response(TypedDict):
    summary : str
    sentiment : str
```

```python
class ConfigureModel:

    def __init__(self, model_name:str = "google/gemma-3-27b-it"):
        self.model_name = model_name
        self.model = None
        self._initiate_model()
    
    def _initiate_model(self):
        print(f"Trying to connect with model {self.model_name}...")
        try:
            llm = HuggingFaceEndpoint(
                repo_id = self.model_name,
                task = 'text-generation',
                huggingfacehub_api_token = os.getenv('HUGGINGFACEHUB_ACCESS_TOKEN')
            )
            self.model = ChatHuggingFace(llm=llm)
            self.model = self.model.with_structured_output(Response)
            print("Connection Successfull !")

        except Exception as e:
            print(f"Error while connecting with model : {e}")
    
    def get_response(self, prompt):
        response = self.model.invoke(prompt)
        print(type(response))
        return response
```

```python
model_configurer = ConfigureModel()
```

```python
response = model_configurer.get_response("Humans are so selfish. They can't be trusted at all. So forgive them but never trust again")
```
