```python
from langchain_huggingface import HuggingFaceEmbeddings
from typing import List, Any
from sklearn.metrics.pairwise import cosine_similarity
```

## Embedding Manager

```python
class EmbeddingsManager:

    def __init__(self, model_name:str = "sentence-transformers/all-MiniLM-L6-v2"):
        self.model_name = model_name 
        self.model = None
        self._initiate_model()

    def _initiate_model(self):
        print(f"Trying to connect to model {self.model_name}...")
        try:
            self.model = HuggingFaceEmbeddings(model_name = self.model_name)
            print("Connected to model successfully !")
        
        except Exception as e:
            print("Error while connecting to embedding model : {e}")
            raise 
    
    def generate_embeddings(self, texts:List[str]):
        print(f"Trying to generate embeddings for {len(texts)} texts...")
        try:
            embeddings = self.model.embed_documents(texts)
            print("Embeddings generated successfully !")
            return embeddings

        except Exception as e:
            print("Error while connecting to embedding model : {e}")
            raise 
```

```python
embeddings_manager = EmbeddingsManager()
```

## Generating Embeddings

```python
texts = ["Iqra Afzal", "is a pretty nice", "human being"]
```

```python
embeddings = embeddings_manager.generate_embeddings(texts)
```

```python
print(embeddings)
```

## Implementing Cosine Similarity

```python
documents = [
    "Iqra Afzal is a good person", 
    "Moon is white",
    "Being nice is a rare trait",
    "I love AI and space exploration",
    "I love decoding maths and physics too"
]
doc_embeddings = embeddings_manager.generate_embeddings(documents)
```

```python
query = "what kind of person iqra is?"
query_embedding = embeddings_manager.generate_embeddings([query])
```

```python
print(cosine_similarity(query_embedding, doc_embeddings))
```
