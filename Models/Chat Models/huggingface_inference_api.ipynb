```python
from langchain_huggingface import ChatHuggingFace, HuggingFaceEndpoint
from dotenv import load_dotenv
import os
```

## Loading API Key

```python
load_dotenv(override=True)
```

```python
print(os.getenv('HUGGINGFACEHUB_ACCESS_TOKEN') is not None)
```

## Configuring ChatHuggingFace

```python
class ConfigureHuggingFace:

    def __init__(self, repo_id:str = "Qwen/Qwen2.5-7B-Instruct", task:str = "text-generation", huggingfacehub_api_token:str = os.getenv('HUGGINGFACEHUB_ACCESS_TOKEN')):
        self.repo_id = repo_id
        self.task = task
        self.huggingfacehub_api_token = huggingfacehub_api_token
        self.model = None
        self._initiate_model()

    # Private method to initiate model
    def _initiate_model(self):
        print(f"Trying to connect to model {self.repo_id}...")
        try:
            llm = HuggingFaceEndpoint(
                repo_id = self.repo_id,
                task = self.task,
                huggingfacehub_api_token = self.huggingfacehub_api_token
            )

            self.model = ChatHuggingFace(llm = llm)
            print("Connected to model successfully !")
        
        except Exception as e:
            print(f"Error while connecting to model {e}")
            raise
    
    # Method to get model response
    def get_response(self, prompt:str):
        print(f"Query sent to the model, waiting for response...")
        try:
            response = self.model.invoke(prompt)
            print("Response generated sucessfully !")
            return response.content
        
        except Exception as e:
            print(f"Error while connecting to model {e}")
            raise
```

```python
hugging_face_configurer = ConfigureHuggingFace("Qwen/Qwen2.5-7B-Instruct", "text-generation", os.getenv('HUGGINGFACEHUB_ACCESS_TOKEN'))
```

## Getting Response

let's get names of top 5 books for self growth

```python
response = hugging_face_configurer.get_response("Tell me names of top 5 books for self growth")
```

```python
print(response)
```

## Let's Print Books Names

```python
import re
from rich.console import Console
from rich.table import Table
```

#### Extracting books names from response

```python
books = re.findall(r'"([^"]+)"\s+by', response.content)
```

#### Printing books names using module 'rich'

```python
console = Console()
books_table = Table()
```

```python
books_table.add_column("Book", style = "yellow")
```

```python
for book in books:
    books_table.add_row(book)
```

```python
console.print(books_table)
```
