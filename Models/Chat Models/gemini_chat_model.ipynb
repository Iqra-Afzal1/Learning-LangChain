```python
from langchain_google_genai import ChatGoogleGenerativeAI
from dotenv import load_dotenv
import os
```

## Loading API Key

```python
load_dotenv(override=True)
```

```python
print(os.getenv('GOOGLE_API_KEY') is not None)
```

## Configuring Model

```python
class ConfigureModel:

    def __init__(self, model_name:str = "gemini-2.5-flash", google_api_key:str = os.getenv('GOOGLE_API_KEY')):
        self.model_name = model_name
        self.google_api_key = google_api_key
        self.model = None
        self._initiate_model()
    
    # Method to initialize model
    def _initiate_model(self):
        print(f"Trying to connect to model {self.model_name}...")
        try:
            self.model = ChatGoogleGenerativeAI(model=self.model_name, google_api_key = self.google_api_key)
            print(f"Connection to model is successfull !")
        
        except Exception as e:
            print(f"Error while connecting to model {e}")
            raise
    
    # Method to get response
    def get_response(self, prompt:str):
        print("Query sent to model, waiting for the response...")
        try:
            response = self.model.invoke(prompt)               
            print("Response generated sucessfully !")
            return response.content
        
        except Exception as e:
            print(f"Error while connecting to model {e}")
            raise
```

```python
model_configurer = ConfigureModel()
```

## Getting Response

```python
response = model_configurer.get_response("How many r are there in word 'strawberry' ? ")
```
