```python
from langchain_groq import ChatGroq
from dotenv import load_dotenv
from langchain_core.output_parsers import StrOutputParser
from langchain_core.prompts import PromptTemplate
import os
```

```python
load_dotenv(override=True)
```

```python
class ConfigureModel:

    def __init__(self, model_name:str="openai/gpt-oss-20b", groq_api_key=os.getenv('GROQ_API_KEY')):
        self.model_name = model_name
        self.groq_api_key = groq_api_key
        self.model = None
        self._initiate_model()
    
    def _initiate_model(self):
        print(f"Trying to connect with model {self.model_name}...")
        try:
            self.model = ChatGroq(
                model = self.model_name,
                temperature = 0.7,
                groq_api_key = self.groq_api_key
            )
            print("Model connection successfull")
        
        except Exception as e:
            print(f"Error: {e}")

    def get_response(self, prompt):
        print(f"prompt sent to model, waiting for response")
        try:
            response = self.model.invoke(prompt)
            return response.content
        
        except Exception as e:
            print(f"Error: {e}")
        
```

```python
model_configurer = ConfigureModel()
```

```python
prompt = PromptTemplate(
    template = "Write a 5 lines summary about the topic {topic}",
    input_variables = ['topic']
)
```

```python
parser = StrOutputParser()
```

## Forming a chain

- In a chain, output of a component sent automatically becomes input of upcoming component
- We pass input to first component

```python
chain = prompt | model_configurer.model | parser
```

```python
result = chain.invoke({'topic' : 'Iqbal Day'})
```

```python
print(result)
```
