```python
from langchain_community.document_loaders import DirectoryLoader, PyPDFLoader
```

```python
loader = DirectoryLoader(
    path="temp_dir",   # path of the directory
    glob = "*.pdf",    # pattern to specify which type of files to load
    loader_cls = PyPDFLoader  # Loader class to specifiy which loader should be used to load fils
)
```

```python
docs = loader.load()
```

```python
print(len(docs))
```

```python
print(docs[0])
```
