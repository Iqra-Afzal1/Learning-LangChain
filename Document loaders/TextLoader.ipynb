```python
from langchain_community.document_loaders import TextLoader
```

- TextLoader loads content from a text file
- We use .load() method to get list of Document objects (A list is returned becuase source may contain multiple pieces of information)
- , but this splitting isn't done intelligently as PyPDFLoader makes a Document of each page of pdf while usually TextLoader keeps single Document in the list
- Each Document object has "page_content" which is actual data and "metadata" which is extra information about the data

```python
loader = TextLoader("temp.txt", encoding="utf-8")  # making loader object
```

```python
docs = loader.load()  # loading data using .load() method
```

```python
print(docs)
print(type(docs))
print(type(docs[0]))
```
